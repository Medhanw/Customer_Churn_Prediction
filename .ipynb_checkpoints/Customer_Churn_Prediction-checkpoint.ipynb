```python

## 5. Feature Engineering and Model Preparation

##In this section, we prepare the cleaned dataset for machine learning. We separate the independent variables from the target variable, encode categorical features, split the data into training and testing sets, and establish a baseline classification model.
```

```python

import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.linear_model import LogisticRegression

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix,
    roc_auc_score
)
```

```python

# Remove customerID because it is an identifier,
# not a useful predictive feature.

X = df.drop(columns=["customerID", "Churn"])

# Convert target into numerical format
y = df["Churn"].map({
    "No": 0,
    "Yes": 1
})

print("Feature shape:", X.shape)
print("Target shape:", y.shape)

print("\nTarget distribution:")
print(y.value_counts())
```

```python

numerical_features = X.select_dtypes(
    include=["int64", "float64"]
).columns.tolist()

categorical_features = X.select_dtypes(
    include=["object", "category"]
).columns.tolist()

print("Numerical Features:", numerical_features)
print("\nCategorical Features:", categorical_features)
```

```python

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)
print("Training samples:", X_train.shape[0])
print("Testing samples:", X_test.shape[0])
```

```python

# Numerical preprocessing
numeric_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

# Categorical preprocessing
categorical_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore"))
])

# Combine both preprocessing pipelines
preprocessor = ColumnTransformer([
    ("num", numeric_pipeline, numerical_features),
    ("cat", categorical_pipeline, categorical_features)
])

# Complete machine learning pipeline
baseline_model = Pipeline([
    ("preprocessor", preprocessor),
    ("classifier", LogisticRegression(
        max_iter=1000,
        class_weight="balanced",
        random_state=42
    ))
])

# Train model
baseline_model.fit(X_train, y_train)

print("Baseline model trained successfully!")
```

```python

# Generate predictions
y_pred = baseline_model.predict(X_test)

# Generate churn probabilities
y_prob = baseline_model.predict_proba(X_test)[:, 1]

# Calculate evaluation metrics
print("BASELINE MODEL PERFORMANCE")
print("-" * 40)

print("Accuracy:", accuracy_score(y_test, y_pred))
print("Precision:", precision_score(y_test, y_pred))
print("Recall:", recall_score(y_test, y_pred))
print("F1 Score:", f1_score(y_test, y_pred))
print("ROC-AUC:", roc_auc_score(y_test, y_prob))

print("\nClassification Report:")
print(classification_report(y_test, y_pred))
```

```python

# ==========================================
# STEP 4: OUTLIER DETECTION USING IQR
# ==========================================

print("OUTLIER DETECTION")
print("=" * 50)

# Numerical columns for outlier analysis
numerical_columns = [
    "tenure",
    "MonthlyCharges",
    "TotalCharges"
]

outlier_results = []

for column in numerical_columns:

    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    outliers = df[
        (df[column] < lower_limit) |
        (df[column] > upper_limit)
    ]

    outlier_results.append({
        "Feature": column,
        "Q1": Q1,
        "Q3": Q3,
        "IQR": IQR,
        "Lower Limit": lower_limit,
        "Upper Limit": upper_limit,
        "Outlier Count": len(outliers)
    })

outlier_df = pd.DataFrame(outlier_results)

print(outlier_df.to_string(index=False))
```

```python

# ==========================================
# STEP 4.2: OUTLIER VISUALIZATION
# ==========================================

import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(15, 5))

for i, column in enumerate(numerical_columns):

    plt.subplot(1, 3, i + 1)

    sns.boxplot(y=df[column])

    plt.title(f"Outlier Analysis - {column}")
    plt.ylabel(column)

plt.tight_layout()

plt.savefig(
    "Images/outlier_analysis.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python


# ==========================================
# FEATURE ENGINEERING
# ==========================================

print("FEATURE ENGINEERING")
print("=" * 50)

# 1. Customer tenure category
df["TenureCategory"] = pd.cut(
    df["tenure"],
    bins=[-1, 12, 24, 48, 72],
    labels=["New", "Medium", "Long", "Loyal"]
)

# 2. Average monthly spending
df["AverageMonthlySpending"] = (
    df["TotalCharges"] / df["tenure"].replace(0, np.nan)
)

# 3. Total number of additional services
service_columns = [
    "OnlineSecurity",
    "OnlineBackup",
    "DeviceProtection",
    "TechSupport",
    "StreamingTV",
    "StreamingMovies"
]

df["AdditionalServicesCount"] = (
    df[service_columns] == "Yes"
).sum(axis=1)

# 4. Monthly charges category
df["ChargesCategory"] = pd.cut(
    df["MonthlyCharges"],
    bins=[-np.inf, 35, 70, 100, np.inf],
    labels=["Low", "Medium", "High", "Very High"]
)

print("New Features Created:")
print([
    "TenureCategory",
    "AverageMonthlySpending",
    "AdditionalServicesCount",
    "ChargesCategory"
])

print("\nUpdated Dataset Shape:", df.shape)

print("\nNew Feature Preview:")
print(
    df[
        [
            "TenureCategory",
            "AverageMonthlySpending",
            "AdditionalServicesCount",
            "ChargesCategory"
        ]
    ].head()
)
```

```python

# ==========================================
# FEATURE AND TARGET SEPARATION
# ==========================================

# Target variable
y = df["Churn"].map({
    "No": 0,
    "Yes": 1
})

# Remove target and customer identifier
X = df.drop(
    columns=["Churn", "customerID"],
    errors="ignore"
)

print("Feature Shape:", X.shape)
print("Target Shape:", y.shape)

print("\nTarget Distribution:")
print(y.value_counts())
```

```python

# ==========================================
# TRAIN TEST SPLIT
# ==========================================

from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training Data:", X_train.shape)
print("Testing Data:", X_test.shape)

print("\nTraining Target Distribution:")
print(y_train.value_counts(normalize=True).round(3))

print("\nTesting Target Distribution:")
print(y_test.value_counts(normalize=True).round(3))
```

```python

# ==========================================
# DATA PREPROCESSING PIPELINE
# ==========================================

from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.impute import SimpleImputer

# Identify numerical and categorical columns
numerical_columns = X_train.select_dtypes(
    include=["int64", "float64"]
).columns.tolist()

categorical_columns = X_train.select_dtypes(
    include=["object", "category", "bool"]
).columns.tolist()

print("Numerical Features:", numerical_columns)
print("\nCategorical Features:", categorical_columns)

# Numerical preprocessing
numerical_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

# Categorical preprocessing
categorical_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore"))
])

# Combine both pipelines
preprocessor = ColumnTransformer([
    ("num", numerical_pipeline, numerical_columns),
    ("cat", categorical_pipeline, categorical_columns)
])

print("\nPreprocessing pipeline created successfully!")
```

```python

# ==========================================
# RECREATE PREPROCESSING PIPELINE
# ==========================================

from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.impute import SimpleImputer

# Identify numerical columns
numerical_columns = X_train.select_dtypes(
    include=["int64", "float64"]
).columns.tolist()

# Identify categorical columns
categorical_columns = X_train.select_dtypes(
    include=["object", "category", "bool"]
).columns.tolist()

# Numerical pipeline
numerical_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

# Categorical pipeline
categorical_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore"))
])

# Combine preprocessing
preprocessor = ColumnTransformer([
    ("num", numerical_pipeline, numerical_columns),
    ("cat", categorical_pipeline, categorical_columns)
])

print("Preprocessor created successfully!")
print("Numerical columns:", numerical_columns)
print("Categorical columns:", categorical_columns)
```

```python

# ==========================================
# MACHINE LEARNING MODEL TRAINING
# ==========================================

from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    classification_report,
    confusion_matrix
)

models = {
    "Logistic Regression": LogisticRegression(
        max_iter=2000,
        class_weight="balanced",
        random_state=42
    ),

    "Random Forest": RandomForestClassifier(
        n_estimators=300,
        max_depth=12,
        min_samples_split=5,
        class_weight="balanced",
        random_state=42,
        n_jobs=-1
    ),

    "Gradient Boosting": GradientBoostingClassifier(
        n_estimators=150,
        learning_rate=0.05,
        max_depth=3,
        random_state=42
    )
}

trained_models = {}
results = []

for name, model in models.items():

    print(f"\nTraining {name}...")

    pipeline = Pipeline([
        ("preprocessor", preprocessor),
        ("classifier", model)
    ])

    pipeline.fit(X_train, y_train)

    y_pred = pipeline.predict(X_test)
    y_prob = pipeline.predict_proba(X_test)[:, 1]

    metrics = {
        "Model": name,
        "Accuracy": accuracy_score(y_test, y_pred),
        "Precision": precision_score(y_test, y_pred, zero_division=0),
        "Recall": recall_score(y_test, y_pred, zero_division=0),
        "F1 Score": f1_score(y_test, y_pred, zero_division=0),
        "ROC-AUC": roc_auc_score(y_test, y_prob)
    }

    results.append(metrics)
    trained_models[name] = pipeline

    print("\n", name)
    print(classification_report(y_test, y_pred))

results_df = pd.DataFrame(results)

print("\nMODEL COMPARISON")
print("=" * 70)
print(results_df.round(4))
```

```python

# ==========================================
# MODEL PERFORMANCE COMPARISON
# ==========================================

results_melted = results_df.melt(
    id_vars="Model",
    var_name="Metric",
    value_name="Score"
)

plt.figure(figsize=(12, 6))

sns.barplot(
    data=results_melted,
    x="Model",
    y="Score",
    hue="Metric"
)

plt.title("Machine Learning Model Performance Comparison")
plt.ylabel("Performance Score")
plt.ylim(0, 1.05)
plt.xticks(rotation=15)
plt.legend(bbox_to_anchor=(1.02, 1), loc="upper left")

plt.tight_layout()

plt.savefig(
    "Images/model_performance_comparison.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python

# ==========================================
# SAVE MODEL COMPARISON RESULTS
# ==========================================

results_df.to_csv(
    "model_comparison_results.csv",
    index=False
)

print("Model comparison results saved successfully!")

# Identify the model with the highest ROC-AUC
best_model_name = results_df.loc[
    results_df["ROC-AUC"].idxmax(),
    "Model"
]

print("\nBest Model Based on ROC-AUC:", best_model_name)
```

```python

# ==========================================
# STEP 1: HYPERPARAMETER TUNING
# ==========================================

from sklearn.model_selection import GridSearchCV
from sklearn.pipeline import Pipeline

from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier

# Define models and parameter grids
tuning_models = {
    "Logistic Regression": (
        LogisticRegression(max_iter=2000, class_weight="balanced"),
        {
            "classifier__C": [0.1, 1, 10],
            "classifier__solver": ["liblinear", "lbfgs"]
        }
    ),

    "Random Forest": (
        RandomForestClassifier(
            random_state=42,
            class_weight="balanced"
        ),
        {
            "classifier__n_estimators": [100, 200],
            "classifier__max_depth": [5, 10, None],
            "classifier__min_samples_split": [2, 5]
        }
    ),

    "Gradient Boosting": (
        GradientBoostingClassifier(random_state=42),
        {
            "classifier__n_estimators": [100, 200],
            "classifier__learning_rate": [0.05, 0.1],
            "classifier__max_depth": [2, 3]
        }
    )
}

best_estimators = {}
tuning_results = []

for name, (model, params) in tuning_models.items():

    print(f"\nTuning {name}...")

    pipeline = Pipeline([
        ("preprocessor", preprocessor),
        ("classifier", model)
    ])

    grid = GridSearchCV(
        estimator=pipeline,
        param_grid=params,
        scoring="roc_auc",
        cv=5,
        n_jobs=-1,
        verbose=1
    )

    grid.fit(X_train, y_train)

    best_estimators[name] = grid.best_estimator_

    tuning_results.append({
        "Model": name,
        "Best CV ROC-AUC": grid.best_score_,
        "Best Parameters": grid.best_params_
    })

print("\nHYPERPARAMETER TUNING COMPLETED")

tuning_df = pd.DataFrame(tuning_results)

print(tuning_df.to_string(index=False))
```

```python

# ==========================================
# STEP 2: TUNED MODEL COMPARISON
# ==========================================

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score
)

tuned_results = []

for name, model in best_estimators.items():

    y_pred = model.predict(X_test)
    y_prob = model.predict_proba(X_test)[:, 1]

    tuned_results.append({
        "Model": name,
        "Accuracy": accuracy_score(y_test, y_pred),
        "Precision": precision_score(y_test, y_pred, zero_division=0),
        "Recall": recall_score(y_test, y_pred, zero_division=0),
        "F1 Score": f1_score(y_test, y_pred, zero_division=0),
        "ROC-AUC": roc_auc_score(y_test, y_prob)
    })

tuned_results_df = pd.DataFrame(tuned_results)

print("TUNED MODEL PERFORMANCE")
print(tuned_results_df.round(4).to_string(index=False))

# Visual comparison
import matplotlib.pyplot as plt
import seaborn as sns

plot_data = tuned_results_df.set_index("Model")[
    ["Accuracy", "Precision", "Recall", "F1 Score", "ROC-AUC"]
]

plt.figure(figsize=(12, 6))

sns.heatmap(
    plot_data,
    annot=True,
    fmt=".3f",
    cmap="YlGnBu"
)

plt.title("Tuned Model Performance Comparison")
plt.tight_layout()

plt.savefig(
    "Images/tuned_model_comparison.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python

# ==========================================
# STEP 3: FINAL MODEL SELECTION
# ==========================================

best_model_name = tuned_results_df.loc[
    tuned_results_df["ROC-AUC"].idxmax(),
    "Model"
]

best_model = best_estimators[best_model_name]

print("SELECTED MODEL:", best_model_name)

# Predictions
y_pred = best_model.predict(X_test)
y_prob = best_model.predict_proba(X_test)[:, 1]

from sklearn.metrics import classification_report

print("\nCLASSIFICATION REPORT")
print(classification_report(
    y_test,
    y_pred,
    target_names=["Stayed", "Churned"]
))

print("Final ROC-AUC:", round(
    roc_auc_score(y_test, y_prob), 4
))
```

```python

# ==========================================
# CONFUSION MATRIX AND ROC CURVE
# ==========================================

import os

from sklearn.metrics import (
    ConfusionMatrixDisplay,
    RocCurveDisplay
)

os.makedirs("Images", exist_ok=True)

# Confusion Matrix
fig, ax = plt.subplots(figsize=(7, 5))

ConfusionMatrixDisplay.from_predictions(
    y_test,
    y_pred,
    display_labels=["Stayed", "Churned"],
    cmap="Blues",
    ax=ax
)

plt.title(f"{best_model_name} - Confusion Matrix")
plt.tight_layout()

plt.savefig(
    "Images/final_confusion_matrix.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ROC Curve
fig, ax = plt.subplots(figsize=(7, 5))

RocCurveDisplay.from_predictions(
    y_test,
    y_prob,
    name=best_model_name,
    ax=ax
)

plt.title(f"{best_model_name} - ROC Curve")
plt.tight_layout()

plt.savefig(
    "Images/final_roc_curve.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python

# ==========================================
# STEP 4: SAVE FINAL MODEL
# ==========================================

import joblib
import os

os.makedirs("Models", exist_ok=True)

model_path = "Models/customer_churn_model.pkl"

joblib.dump(best_model, model_path)

# Save model information
model_info = {
    "model_name": best_model_name,
    "roc_auc": roc_auc_score(y_test, y_prob),
    "accuracy": accuracy_score(y_test, y_pred),
    "precision": precision_score(y_test, y_pred),
    "recall": recall_score(y_test, y_pred),
    "f1_score": f1_score(y_test, y_pred)
}

joblib.dump(
    model_info,
    "Models/model_metadata.pkl"
)

print("FINAL MODEL SAVED SUCCESSFULLY!")
print("Model:", best_model_name)
print("Location:", model_path)
print("ROC-AUC:", round(model_info["roc_auc"], 4))
```

```python

# ==========================================
# CUSTOMER RISK SEGMENTATION
# ==========================================

def assign_risk(probability):

    if probability < 0.30:
        return "Low"

    elif probability <= 0.60:
        return "Medium"

    else:
        return "High"


customer_risk["Risk_Category"] = (
    customer_risk["Churn_Probability"]
    .apply(assign_risk)
)

print("CUSTOMER RISK DISTRIBUTION")

print(
    customer_risk["Risk_Category"]
    .value_counts()
)

print("\nSample Customer Risk:")
print(customer_risk.head(10))
```

```python

# ==========================================
# STEP 7: RISK DISTRIBUTION ANALYSIS
# ==========================================

risk_summary = customer_risk.groupby(
    "Risk_Category"
).agg(
    Total_Customers=("CustomerID", "count"),
    Average_Churn_Probability=("Churn_Probability", "mean")
).reset_index()

risk_summary["Customer_Percentage"] = (
    risk_summary["Total_Customers"]
    / len(customer_risk) * 100
)

print("CUSTOMER RISK SUMMARY")
print(risk_summary.round(2).to_string(index=False))
```

```python

# ==========================================
# STEP 8: ACTUAL CHURN VS PREDICTED RISK
# ==========================================

risk_validation = pd.crosstab(
    customer_risk["Risk_Category"],
    customer_risk["Actual_Churn"],
    normalize="index"
) * 100

print("CHURN RATE WITHIN EACH RISK CATEGORY")
print(risk_validation.round(2))

# Visualization
risk_validation.plot(
    kind="bar",
    stacked=True,
    figsize=(9, 5)
)

plt.title("Actual Churn Distribution Across Risk Categories")
plt.xlabel("Predicted Risk Category")
plt.ylabel("Percentage of Customers")

plt.legend(title="Actual Churn")
plt.xticks(rotation=0)
plt.tight_layout()

plt.savefig(
    "Images/risk_validation.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python

# ==========================================
# STEP 9: RETENTION RECOMMENDATION ENGINE
# ==========================================

# Add customer information
customer_risk = customer_risk.merge(
    df.drop(columns=["Churn"]),
    left_on="CustomerID",
    right_on="customerID",
    how="left"
)

def retention_recommendation(row):

    recommendations = []

    if row["Risk_Category"] == "High":
        recommendations.append(
            "Priority retention outreach"
        )

    if row["Contract"] == "Month-to-month":
        recommendations.append(
            "Offer long-term contract benefits"
        )

    if row["OnlineSecurity"] == "No":
        recommendations.append(
            "Offer online security package"
        )

    if row["TechSupport"] == "No":
        recommendations.append(
            "Offer technical support package"
        )

    if row["PaymentMethod"] == "Electronic check":
        recommendations.append(
            "Offer alternative payment options"
        )

    if not recommendations:
        recommendations.append(
            "Continue regular customer engagement"
        )

    return "; ".join(recommendations)


customer_risk["Retention_Recommendations"] = (
    customer_risk.apply(
        retention_recommendation,
        axis=1
    )
)

print(
    customer_risk[
        [
            "CustomerID",
            "Churn_Probability",
            "Risk_Category",
            "Retention_Recommendations"
        ]
    ].head(15).to_string(index=False)
)
```

```python

# ==========================================
# STEP 10: EXECUTIVE DASHBOARD METRICS
# ==========================================

total_customers = len(customer_risk)

high_risk = (
    customer_risk["Risk_Category"] == "High"
).sum()

medium_risk = (
    customer_risk["Risk_Category"] == "Medium"
).sum()

low_risk = (
    customer_risk["Risk_Category"] == "Low"
).sum()

average_probability = (
    customer_risk["Churn_Probability"].mean() * 100
)

dashboard_metrics = {
    "Total Customers": total_customers,
    "High Risk Customers": high_risk,
    "Medium Risk Customers": medium_risk,
    "Low Risk Customers": low_risk,
    "Average Predicted Churn Probability": round(
        average_probability, 2
    )
}

print("CUSTOMER CHURN EXECUTIVE DASHBOARD")
print("=" * 45)

for metric, value in dashboard_metrics.items():
    print(f"{metric}: {value}")

# Export dashboard data
customer_risk.to_csv(
    "customer_risk_report.csv",
    index=False
)

pd.DataFrame(
    [dashboard_metrics]
).to_csv(
    "dashboard_metrics.csv",
    index=False
)

print("\nDashboard datasets exported successfully!")
```
